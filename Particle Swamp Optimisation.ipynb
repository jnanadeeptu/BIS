```python
import random

# Items
weights = [2, 3, 4, 5]
profits = [20, 30, 40, 50]

capacity = 7

POPULATION_SIZE = 10
GENERATIONS = 50
MUTATION_RATE = 0.1


# Calculate fitness
def fitness(chromosome):

    total_weight = 0
    total_profit = 0

    for i in range(len(chromosome)):
        if chromosome[i] == 1:
            total_weight += weights[i]
            total_profit += profits[i]

    # Invalid solution
    if total_weight > capacity:
        return 0

    return total_profit


# Create initial population
population = []

for _ in range(POPULATION_SIZE):
    chromosome = [
        random.randint(0, 1)
        for _ in range(len(weights))
    ]
    population.append(chromosome)


# Genetic Algorithm
for generation in range(GENERATIONS):

    # Selection
    population.sort(key=fitness, reverse=True)

    parents = population[:5]

    new_population = parents.copy()

    # Crossover
    while len(new_population) < POPULATION_SIZE:

        parent1 = random.choice(parents)
        parent2 = random.choice(parents)

        crossover_point = random.randint(1, len(weights) - 1)

        child = (
            parent1[:crossover_point]
            + parent2[crossover_point:]
        )

        # Mutation
        for i in range(len(child)):
            if random.random() < MUTATION_RATE:
                child[i] = 1 - child[i]

        new_population.append(child)

    population = new_population


# Best solution
best = max(population, key=fitness)

total_weight = 0
total_profit = 0

for i in range(len(best)):
    if best[i] == 1:
        total_weight += weights[i]
        total_profit += profits[i]


print("Best chromosome:", best)
print("Total weight:", total_weight)
print("Maximum profit:", total_profit)
```

```python
import random

# Fitness function
def fitness(x):
    return x ** 2


# Parameters
POP_SIZE = 10
GENERATIONS = 20
MUTATION_RATE = 0.1

# Initial population
population = [random.randint(0, 31) for _ in range(POP_SIZE)]

for generation in range(GENERATIONS):

    # Selection
    population.sort(key=fitness, reverse=True)
    parents = population[:5]

    new_population = parents.copy()

    # Crossover
    while len(new_population) < POP_SIZE:

        p1 = random.choice(parents)
        p2 = random.choice(parents)

        child = (p1 + p2) // 2

        # Mutation
        if random.random() < MUTATION_RATE:
            child += random.randint(-2, 2)

        # Keep x within range
        child = max(0, min(31, child))

        new_population.append(child)

    population = new_population


# Best solution
best = max(population, key=fitness)

print("Best x:", best)
print("Maximum value:", fitness(best))
```

```python
import random

# Objective function
def fitness(x):
    return x ** 2


# Parameters
NUM_PARTICLES = 10
ITERATIONS = 50
W = 0.5
C1 = 1.5
C2 = 1.5

# Initialize particles
particles = []

for _ in range(NUM_PARTICLES):
    position = random.uniform(-10, 10)
    velocity = random.uniform(-1, 1)

    particles.append({
        "position": position,
        "velocity": velocity,
        "best_position": position,
        "best_fitness": fitness(position)
    })


# Find global best
global_best = min(particles, key=lambda p: p["best_fitness"])
global_best_position = global_best["best_position"]


# PSO algorithm
for _ in range(ITERATIONS):

    for particle in particles:

        r1 = random.random()
        r2 = random.random()

        # Update velocity
        particle["velocity"] = (
            W * particle["velocity"]
            + C1 * r1 * (
                particle["best_position"]
                - particle["position"]
            )
            + C2 * r2 * (
                global_best_position
                - particle["position"]
            )
        )

        # Update position
        particle["position"] += particle["velocity"]

        # Calculate fitness
        current_fitness = fitness(particle["position"])

        # Update personal best
        if current_fitness < particle["best_fitness"]:
            particle["best_position"] = particle["position"]
            particle["best_fitness"] = current_fitness

    # Update global best
    best_particle = min(
        particles,
        key=lambda p: p["best_fitness"]
    )

    if best_particle["best_fitness"] < fitness(global_best_position):
        global_best_position = best_particle["best_position"]


print("Best position:", global_best_position)
print("Minimum value:", fitness(global_best_position))
```
