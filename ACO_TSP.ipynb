```python
import random

# Distance between cities
dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(dist)
ants = 4
iterations = 50
alpha = 1
beta = 2
rho = 0.5
Q = 100

# Initialize pheromones
pheromone = [[1 for j in range(n)] for i in range(n)]

best_tour = None
best_distance = float('inf')

for _ in range(iterations):

    tours = []
    lengths = []

    # Construct solutions
    for ant in range(ants):
        tour = [random.randint(0, n - 1)]

        while len(tour) < n:
            current = tour[-1]
            unvisited = [c for c in range(n) if c not in tour]

            weights = [
                pheromone[current][c] ** alpha *
                (1 / dist[current][c]) ** beta
                for c in unvisited
            ]

            next_city = random.choices(
                unvisited, weights=weights
            )[0]

            tour.append(next_city)

        # Return to starting city
        length = sum(
            dist[tour[i]][tour[(i + 1) % n]]
            for i in range(n)
        )

        tours.append(tour)
        lengths.append(length)

        # Record best tour
        if length < best_distance:
            best_distance = length
            best_tour = tour[:]

    # Evaporate pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # Update pheromones
    for tour, length in zip(tours, lengths):
        for i in range(n):
            a = tour[i]
            b = tour[(i + 1) % n]
            pheromone[a][b] += Q / length
            pheromone[b][a] += Q / length

# Output result
print("Best Tour:", best_tour + [best_tour[0]])
print("Shortest Distance:", best_distance)
```
